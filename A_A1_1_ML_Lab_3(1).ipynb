```python
X = dl.iloc[:, : -1]
y = dl.iloc[:, -1]
```

```python
import pandas as pd
import numpy as np


print("Enjoy Sports Dataset:")
print(dl)


X = dl.iloc[:, :-1].values
y = dl.iloc[:, -1].values

def find_s(X, y):
    hypothesis = None
    print("Initial Hypothesis:",hypothesis)
    for i in range(len(X)):
      if y[i] == "Yes":
        if hypothesis is None:
          hypothesis = X[i].copy()
        else:
          for j in range(len(hypothesis)):
            if hypothesis[j] != X[i][j]:
              hypothesis[j] = "?"
    print(f"\nAfter Training Example {i+1}")
    print("Hypothesis:", hypothesis)
    return hypothesis

final_hypothesis = find_s(X, y)
print("\nFinal Hypothesis:", final_hypothesis)
```

```python
def candidate_elimination(concepts, target):
  specific = concepts[0].copy()
  general = [["?" for i in range(len(specific))]]
  print("\nInitial S:", specific)
  print("\nInitial G:", general)
  for i, h in enumerate(concepts):
    if target[i] == "Yes":
      for x in range(len(specific)):
        if h[x] != specific[x]:
          specific[x] = "?"
        else:
          for x in range(len(specific)):
            if h[x] != specific[x]:
              general.append(["?" if j != x else specific[x] for j in range(len(specific))])
    print("\nExample", i+1)
    print("S = ", specific)
    print("G = ", general)
  return specific, general
S, G = candidate_elimination(X, y)
print("\nFinal Specific Boundary")
print(S)
print("\nFinal General Boundary")
print(G)
```

```python
print("========== HYPOTHESIS COMPARISON ==========")

print("\nFind-S Final Hypothesis:")
print(final_hypothesis)

print("\nCandidate Elimination Final Specific Boundary:")
print(S)

print("\nCandidate Elimination Final General Boundary:")
print(G)

print("\nComparison:")
print("1. Find-S produces a single maximally specific hypothesis.")
print("2. Candidate Elimination produces both specific and general boundaries.")
print("3. The Find-S hypothesis represents one hypothesis consistent with the positive examples.")
print("4. Candidate Elimination maintains a set of hypotheses representing the possible version space.")
```
